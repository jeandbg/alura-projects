```python
df
```

```python
df.info()
```

```python
df.set_index('País', inplace=True)    # País virou meu índice
```

```python
anos = list(map(str, range(1980,2014)))
```

```python
anos
```

```python
brasil = df.loc['Brasil',anos]
```

```python
brasil
```

```python
brasil_dict = {'ano': brasil.index.tolist(), 'imigrantes': brasil.values.tolist()}
dados_brasil = pd.DataFrame(brasil_dict)
```

```python
dados_brasil
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,4))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])

plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])    # xticks em strings
plt.yticks([500, 1000, 1500, 2000, 2500, 3000])                         # yticks em ints... vai entender.
#                                                                       # esses yticks aparecem por default, então não precisa. mas botei pra exemplificar

plt.title('Imigração do Brasil para o Canadá')      # a professora sugeriu botar antes dos X e Y ticks, mas eu estou testando e aparentemente
#                                                   # os elementos são 'independentes', posso colocar na ordem que eu quiser.
plt.xlabel('Ano')
plt.ylabel('Número de Imigrantes')

plt.show()   # plt.show() é igual o return: interrompe o código (então tem que ser a última linha da célula)
#            # plt.show() é um método que some com os textos de objetos e locais de memória que aparecem se a gente não usa ele.
```
