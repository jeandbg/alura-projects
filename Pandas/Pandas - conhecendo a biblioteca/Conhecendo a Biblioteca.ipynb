## Filtros

### **1. Apartamentos que possuem `1 quarto` e `aluguel menor que 1200`**



```python
# poderia usar .query(), mas a professora prefere mostrar maneiras mais manuais (por questão de exercício)

df.Quartos == 1
```

```python
selecao1 = df.Quartos == 1
df[selecao1]        # é uma propriedade do Pandas receber uma series booleana desta forma e retornar apenas os valores True
```

```python
selecao2 = df.Valor < 1200
df[selecao2]        # então, na selecao1, fizemos a condição quartos =1; aqui, fazemos Valor < 1200
#                   # agora temos que pegar a intersecção desses dois conjuntos 
```

```python
selecao_final = (selecao1) & (selecao2)
df[selecao_final]
```

```python
df_1 = df[selecao_final]
```

#### Então, antes de seguir em frente, quero testar um pouco as possibilidades.

```python
# não dá pra usar 'and' nem 'not', porque "The truth value of a Series is ambiguous". 
# Daí ele tá considerando a variável series e não cada um dos valores do conjunto. 

# mas dá pra fazer De Morgan pra achar o mesmo resultado usando os sinais trocados.

selecao_final != (selecao1) | (selecao2)
df[selecao_final]
```

```python
# outra coisa. A separação dos argumentos é simplesmente por uma questão didática, mas dá pra fazer

selecao_conjunta = (df['Quartos'] == 1) & (df.Valor < 1200)
df[selecao_conjunta]
```

```python
# por fim, fazendo a busca utilizando .query()

df.query('Quartos == 1 & Valor < 1200')
```

## Criando colunas categóricas

```python
# se eu somo strings o Pandas concatena, bem estilo python mesmo.

dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro']
dados.head()
```

```python
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro'] + ' com ' + \
                     dados['Quartos'].astype(str) + ' quarto(s) ' + \
                     ' e ' + dados['Vagas'].astype(str) + ' vaga(s) de garagem.'    # o Pandas não deixa concatenar str com int/floats
dados.head()                                                                        # tem que usar "astype porque se não dá erro"
```
